> # Pauli X gate

1. X|0> = |1>

2. X|1> = |0>

```python
qc=QuantumCircuit(1)
qc.x(0)
qc.draw(output="mpl")
```

```python
array_to_latex(Operator(qc))
```

```python
array_to_latex(Operator(XGate()).data)
```

```python
op_X=Pauli("X")
array_to_latex(op_X)
```

> **Circuit Identities**
- SXSdg= Y 
- SYSdg=-X
- SZSdg= Z

- XXX= X --------- YXY=-X -------- ZXZ=-X
- XYX= Y --------- YYY= Y -------- ZYZ=-Y
- XZX= Z --------- YZY=-Z -------- ZZZ= Z

- HXH= Z
- HYH= -Y
- HZH= X
