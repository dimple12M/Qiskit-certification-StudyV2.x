> # plot_histogram :
Plot a histogram of input counts data.

REFERENCE : https://quantum.cloud.ibm.com/docs/en/api/qiskit/qiskit.visualization.plot_histogram

```python
from qiskit.visualization import plot_histogram
counts_data = {'00': 499, '11': 501}
plot_histogram(counts_data)
```

> more than 1 measurement counts

```python
counts1 = {'00': 499, '11': 501}
counts2 = {'00': 511, '11': 489}

data = [counts1, counts2]
plot_histogram(data)
```

> OPTIONs in HISTOGRAM

> legend: Provides a label for the executions. It takes a list of strings used to label each execution's results.

```python
 plot_histogram([counts1,counts2],legend=["first", "second"])
```

> sort: Adjusts the order of the bars in the histogram.

> bar_labels: Adjusts whether labels are printed above the bars

```python
 plot_histogram([counts1,counts2],legend=["first", "second"], sort="desc")
```

> Arguments : title, legend,figsize,color,ax and filename

```python
 plot_histogram([counts1,counts2],
                legend=["first", "second"], 
                title="Experiment counts",
                figsize=(5,5),
                color=["blue","red"])
```

> Save the counts histogram file
NOTE: 2 ways to perform

```python
plot_histogram(counts1,filename="counts_fig1.jpg")
```

```python
hist = plot_histogram(counts1)
hist.savefig('new_hist.png')
```

> # Distribution Visualizations
functions that visualize sampled distributions

Plot a distribution from input sampled data.

REFERENCE: https://quantum.cloud.ibm.com/docs/en/api/qiskit/qiskit.visualization.plot_distribution

```python
from qiskit.visualization import plot_distribution
plot_distribution([counts1,counts2],
                legend=["first", "second"], 
                title="Experiment counts",
                color=["blue","red"])
```

```python
counts = {'001': 596, '011': 211, '010': 50, '000': 117, '101': 33, '111': 8,
        '100': 6, '110': 3}

plot_distribution(counts)
```

```python
plot_distribution(counts,sort="desc")
```

```python
plot_distribution(counts,sort="value_desc")
```
