> # plot quantum states

REFERENCE : https://quantum.cloud.ibm.com/docs/en/guides/plot-quantum-states

```python
from math import pi
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector


qc=QuantumCircuit(2,2)
qc.h(0)
qc.x(1)
qc.draw(output="mpl")
```

```python
psi = Statevector(qc)
print(psi)
psi.draw("latex")
```

**NOTE:** Statevectors default to "ket notation"

> ## plot_bloch_vector

REFERENCE: https://quantum.cloud.ibm.com/docs/en/api/qiskit/qiskit.visualization.plot_bloch_vector

Plot a Bloch sphere with the specified coordinates, that can be given in both cartesian and spherical systems.

- plots a quantum system
- works for a single qubit
- takes as input the Bloch vector.

The Bloch vector is defined as [𝑥=Tr[𝑋𝜌],𝑦=Tr[𝑌𝜌],𝑧=Tr[𝑍𝜌]]

where 𝑋, 𝑌and 𝑍 are the Pauli operators for a single qubit and ρ is the state matrix.
input is bloch i.e array of elements
- cartesian: [x,y,z]
- spherical: [r,theta,phi]
- default is cartesian(IMP)
- coord_type : cartesian or spherical


```python
from qiskit.visualization import plot_bloch_vector

plot_bloch_vector([0,1,0], title="New Bloch Sphere")
```

```python
plot_bloch_vector([1,0,0], title="New Bloch Sphere")
```

```python
plot_bloch_vector([0,0,1])
```

```python
plot_bloch_vector([1,0,1])
```

```python
plot_bloch_vector([0,0,-1])
```

```python
from math import sqrt
plot_bloch_vector([0,0,1/sqrt(2)])
```

```python
from math import pi
plot_bloch_vector([pi/2,0,0])
```

```python
import numpy as np
v=[1/sqrt(2),0,0]

normalized_v = v/np.linalg.norm(v)

print(normalized_v)
plot_bloch_vector(v)
```

> ## Spherical coordinates

```python
plot_bloch_vector([1,pi/2,pi/2],coord_type="spherical")
```

r, theta, phi = x,y,z

- x = r * np.sin(theta) * np.cos(phi)
- y = r * np.sin(theta) * np.sin(phi)
- z = r * np.cos(theta)

```python
r, theta, phi=1,pi/2,pi/2

x = r * np.sin(theta) * np.cos(phi)
y = r * np.sin(theta) * np.sin(phi)
z = r * np.cos(theta)
```

```python
print("x= " , x)
print("y= " , y)
print("z= " , z)

plot_bloch_vector([x,y,z])
```
