> ## PhaseGate : p

- Single-qubit rotation about the Z axis.

- diagonal gate

![](./phase_gate2.png)

```python
from qiskit import QuantumCircuit
import numpy as np
qc = QuantumCircuit(1)

qc.p(np.pi/3, 0)  # P(pi/3)

qc.draw("mpl")
```

```python
from qiskit.quantum_info import Operator
from qiskit.visualization import array_to_latex
array_to_latex(Operator(qc))
```

```python
from qiskit.quantum_info import Statevector

psi = Statevector(qc)
psi.draw("latex")
```

> #  P (π/3) |0>= |0>

your circuit starts with |0>
, applying (π/3) first does not produce an observable phase, because there is no |1> component to which the phase can be applied.

> # Q “Can we directly measure the phase of a quantum state?”

<h4>

Global phase cannot be measured, and relative phase cannot be observed directly in the computational basis. 

Relative phase can be determined through interference, such as applying a Hadamard gate before measurement.

</h4>

> # Applying the sequence of quantum gates H (Hadamard), then P(π/3) (Phase shift), and then H performs a rotation around the X-axis of the Bloch sphere.

Mathematically, the identity H P(theta) H = e^{i\theta/2} R_x(theta)) applies. 

For θ = π/3, this gives an \(R_x(\pi/3)\) rotation up to a global phase.

```python
qc = QuantumCircuit(1)

qc.h(0)
qc.p(np.pi/3, 0)
qc.h(0)


qc.draw("mpl")
```

```python
psi = Statevector(qc)
psi.draw("latex")
```

```python
qc = QuantumCircuit(1)
qc.rx(np.pi/3,0)
qc.draw("mpl")
```

```python
psi = Statevector(qc)
psi.draw("latex")
```
